```python
import pennylane as qp

dev = qp.device("default.qubit", wires= 4)

def Uf():
    # The oracle in charge of encoding a hidden "a" value.
    qp.CNOT(wires= [1, 3])
    qp.CNOT(wires= [2, 3])
    
@qp.set_shots(1)
@qp.qnode(dev)
def circuit0():
    """Circuit used to derive a0"""
    
    # Initialize x = [1, 0, 0]
    qp.PauliX(wires = 0)
    
    # Apply our oracle
    
    Uf()
    
    # We measure the last qubit
    return qp.sample(wires=3)

@qp.set_shots(1)
@qp.qnode(dev)
def circuit1():
    """ Circuit used to derive a1"""
     # Initialize x = [0,1,0]
    qp.PauliX(wires = 1)

    # We apply our oracle
    Uf()

    # We measure the last qubit
    return qp.sample(wires = 3)

@qp.set_shots(1)
@qp.qnode(dev)
def circuit2():
    # Circuit used to derive a2
    # Initialize x = [0,0,1]
    qp.PauliX(wires = 2)

    # We apply our oracle
    Uf()

    # We measure the last qubit
    return qp.sample(wires = 3)

# We run for x = [1,0,0]
a0 = circuit0()

# We run for x = [0,1,0]
a1 = circuit1()

# We run for x = [0,0,1]
a2 = circuit2()

print(f"The value of 'a' is [{a0},{a1},{a2}]")
    
```

# Quantum implementation of Bernstein-Vazirani

```python
@qp.set_shots(1)
@qp.qnode(dev)
def circuit():
    
    # We initialize to |0001>
    qp.PauliX(wires = 3)
    
    # We run the Hadamards
    for i in range(4):
        qp.Hadamard(wires=i)
    
    # We apply our function
    Uf()
    
    # We run the Hadamards
    for i in range(4):
        qp.Hadamard(wires=i)
    
    # We measure the first 3 wires
    return qp.sample(wires=[0,1,2])

a = circuit()

print(f"The value of a is {a}")
```

# Qutrit implementation

```python
dev = qp.device("default.qutrit", wires=4)

def Uf():
    # The oracle in charge of encoding a hidden "a" value.
    qp.TAdd(wires= [1, 3])
    qp.TAdd(wires= [1, 3])
    qp.TAdd(wires= [2, 3])
    
@qp.set_shots(1)
@qp.qnode(dev)
def circuit0():
    # Initialize x to [1,0,0]
    qp.TShift(wires=0)
    
    # We apply our oracle
    Uf()
    
    # We measure the last qubit
    return qp.sample(wires=3)

@qp.set_shots(1)
@qp.qnode(dev)
def circuit1():
    # Initialize x to [1,0,0]
    qp.TShift(wires=1)
    
    # We apply our oracle
    Uf()
    
    # We measure the last qubit
    return qp.sample(wires=3)

@qp.set_shots(1)
@qp.qnode(dev)
def circuit2():
    # Initialize x to [1,0,0]
    qp.TShift(wires=2)
    
    # We apply our oracle
    Uf()
    
    # We measure the last qubit
    return qp.sample(wires=3)

# Run to obtain the three trits of a
a0 = circuit0()
a1 = circuit1()
a2 = circuit2()

print(f"The value of a is [{a0}, {a1}, {a2}]") 
```

```python
@qp.set_shots(1)
@qp.qnode(dev)
def circuit():
    # We initialize to |0001>
    qp.TShift(wires=3)
    
    # We run the THadamard
    for i in range(4):
        qp.THadamard(wires=i)
    
    # We run the oracle
    Uf()
    
    # We run the THadamards
    for i in range(4):
        qp.THadamard(i)
    
    # We measure the qutrit
    return qp.sample(wires = range(3))

a = circuit()
print(f"The value of a is {a}")
```
